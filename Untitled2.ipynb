```python
import tensorflow as tf
from tensorflow.keras import layers, Model
import os
import time
from IPython import display
import matplotlib.pyplot as plt

# ==========================================
# 1. DATA PIPELINE (Bulletproof Version)
# ==========================================
ART_DIR = '/Users/aahannayak/artwork_dataset/pixel_art'
SKETCH_DIR = '/Users/aahannayak/artwork_dataset/sketches'

def get_valid_files(directory):
    valid_exts = ('.jpg', '.jpeg', '.png')
    files = [os.path.join(directory, f) for f in os.listdir(directory) if f.lower().endswith(valid_exts)]
    return sorted(files)

sketch_paths = get_valid_files(SKETCH_DIR)
art_paths = get_valid_files(ART_DIR)

def load_and_preprocess(sketch_path, art_path):
    def process(path):
        img = tf.io.read_file(path)
        # Using decode_png/jpeg directly prevents the dynamic shape crash
        img = tf.io.decode_image(img, channels=3, expand_animations=False)
        img.set_shape([None, None, 3]) # Force TensorFlow to know the channels
        
        img = tf.cast(img, tf.float32)
        img = tf.image.resize(img, [64, 64])
        # Normalize to [-1, 1] for the tanh generator
        return (img / 127.5) - 1.0
    return process(sketch_path), process(art_path)

@tf.function
def augment_images(sketch, art):
    sketch = tf.image.resize(sketch, [72, 72], method='nearest')
    art = tf.image.resize(art, [72, 72], method='nearest')
    
    stacked = tf.concat([sketch, art], axis=-1)
    cropped = tf.image.random_crop(stacked, size=[64, 64, 6])
    sketch_aug, art_aug = cropped[:, :, :3], cropped[:, :, 3:]
    
    # Use tf.cond for safe graph execution during dataset mapping
    flip_cond = tf.random.uniform(()) > 0.5
    sketch_aug = tf.cond(flip_cond, lambda: tf.image.flip_left_right(sketch_aug), lambda: sketch_aug)
    art_aug = tf.cond(flip_cond, lambda: tf.image.flip_left_right(art_aug), lambda: art_aug)
    
    return sketch_aug, art_aug

BATCH_SIZE = 4 
full_dataset = tf.data.Dataset.from_tensor_slices((sketch_paths, art_paths))
full_dataset = full_dataset.map(load_and_preprocess, num_parallel_calls=tf.data.AUTOTUNE)

test_size = max(1, int(len(sketch_paths) * 0.2))
train_dataset = full_dataset.skip(test_size).map(augment_images).shuffle(100).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
test_dataset = full_dataset.take(test_size).batch(BATCH_SIZE)

# ==========================================
# 2. THE ARCHITECTURES
# ==========================================
def build_unet_generator():
    inputs = layers.Input(shape=[64, 64, 3]) 
    
    # ENCODER
    e1 = layers.Conv2D(64, 4, strides=2, padding='same', activation='relu')(inputs)
    e2 = layers.Conv2D(128, 4, strides=2, padding='same')(e1)
    e2 = layers.BatchNormalization()(e2)
    e2 = layers.LeakyReLU()(e2)
    e3 = layers.Conv2D(256, 4, strides=2, padding='same')(e2)
    e3 = layers.BatchNormalization()(e3)
    e3 = layers.LeakyReLU()(e3)
    
    # DECODER (With Skip Connections)
    d1 = layers.Conv2DTranspose(128, 4, strides=2, padding='same')(e3)
    d1 = layers.BatchNormalization()(d1)
    d1 = layers.ReLU()(d1)
    d1 = layers.Concatenate()([d1, e2]) # Skip connection from e2
    
    d2 = layers.Conv2DTranspose(64, 4, strides=2, padding='same')(d1)
    d2 = layers.BatchNormalization()(d2)
    d2 = layers.ReLU()(d2)
    d2 = layers.Concatenate()([d2, e1]) # Skip connection from e1
    
    last = layers.Conv2DTranspose(3, 4, strides=2, padding='same', activation='tanh')(d2)
    return Model(inputs=inputs, outputs=last, name="Generator")

def build_conditional_discriminator():
    # Conditional Inputs: It must see the sketch AND the art
    inp = layers.Input(shape=[64, 64, 3], name='sketch_input')
    tar = layers.Input(shape=[64, 64, 3], name='art_target')
    
    x = layers.Concatenate()([inp, tar]) # Shape becomes 64x64x6
    
    x = layers.Conv2D(64, 4, strides=2, padding='same')(x)
    x = layers.LeakyReLU()(x)
    x = layers.Dropout(0.3)(x)
    
    x = layers.Conv2D(128, 4, strides=2, padding='same')(x)
    x = layers.BatchNormalization()(x)
    x = layers.LeakyReLU()(x)
    
    x = layers.Flatten()(x)
    last = layers.Dense(1)(x) 
    return Model(inputs=[inp, tar], outputs=last, name="Discriminator")

generator = build_unet_generator()
discriminator = build_conditional_discriminator() 

# ==========================================
# 3. LOSS, OPTIMIZERS & TRAINING
# ==========================================
# beta_1=0.5 is standard for GAN stability
generator_optimizer = tf.keras.optimizers.Adam(2e-4, beta_1=0.5)
discriminator_optimizer = tf.keras.optimizers.Adam(2e-4, beta_1=0.5)
cross_entropy = tf.keras.losses.BinaryCrossentropy(from_logits=True)

@tf.function
def train_step(input_image, target_image):
    with tf.GradientTape() as gen_tape, tf.GradientTape() as disc_tape:
        generated_images = generator(input_image, training=True)
        
        # Discriminator evaluates REAL pairs and FAKE pairs
        real_output = discriminator([input_image, target_image], training=True)
        fake_output = discriminator([input_image, generated_images], training=True)
        
        # Generator Loss: Fool the discriminator + minimize pixel distance (L1)
        gen_bce_loss = cross_entropy(tf.ones_like(fake_output), fake_output)
        l1_loss = tf.reduce_mean(tf.abs(target_image - generated_images))
        total_gen_loss = gen_bce_loss + (100 * l1_loss) # L1 is weighted heavily
        
        # Discriminator Loss: Detect reals as 1, fakes as 0
        disc_loss = cross_entropy(tf.ones_like(real_output), real_output) + \
                    cross_entropy(tf.zeros_like(fake_output), fake_output)

    grad_gen = gen_tape.gradient(total_gen_loss, generator.trainable_variables)
    grad_disc = disc_tape.gradient(disc_loss, discriminator.trainable_variables)
    
    generator_optimizer.apply_gradients(zip(grad_gen, generator.trainable_variables))
    discriminator_optimizer.apply_gradients(zip(grad_disc, discriminator.trainable_variables))

def generate_and_show_images(model, test_input, tar):
    prediction = model(test_input, training=False)
    plt.figure(figsize=(9, 3))
    
    display_list = [test_input[0], tar[0], prediction[0]]
    title = ['Input Sketch', 'Ground Truth Art', 'Predicted Art']
    
    for i in range(3):
        plt.subplot(1, 3, i+1)
        plt.title(title[i])
        # Rescale [-1, 1] back to [0, 1] for display
        plt.imshow(display_list[i] * 0.5 + 0.5)
        plt.axis('off')
    plt.show()

def train(dataset, test_dataset, epochs):
    # Grab a fixed test batch to watch it evolve over time
    example_input, example_target = next(iter(test_dataset.take(1)))
    
    for epoch in range(epochs):
        start = time.time()
        
        for sketch_batch, art_batch in dataset:
            train_step(sketch_batch, art_batch)
            
        display.clear_output(wait=True)
        generate_and_show_images(generator, example_input, example_target)
        print(f'Epoch {epoch+1} completed in {time.time()-start:.2f} sec')

# ==========================================
# 4. EXECUTE TRAINING
# ==========================================
print("Starting Training Loop...")
train(train_dataset, test_dataset, epochs=1000)
```

```python
import tensorflow as tf
from tensorflow.keras import layers, Model
import matplotlib.pyplot as plt
import numpy as np
import os
import time
from IPython import display

# ==========================================
# 1. DATA PIPELINE & HYPERPARAMETERS
# ==========================================
# We only need the sketches to teach it how to draw from scratch
SKETCH_DIR = '/Users/aahannayak/artwork_dataset/sketches'

def get_valid_files(directory):
    valid_exts = ('.jpg', '.jpeg', '.png')
    files = [os.path.join(directory, f) for f in os.listdir(directory) if f.lower().endswith(valid_exts)]
    return sorted(files)

sketch_paths = get_valid_files(SKETCH_DIR)

def load_and_preprocess(path):
    img = tf.io.read_file(path)
    img = tf.io.decode_image(img, channels=3, expand_animations=False)
    img.set_shape([None, None, 3])
    img = tf.cast(img, tf.float32)
    img = tf.image.resize(img, [64, 64])
    return (img / 127.5) - 1.0 # Scale to [-1, 1]

BATCH_SIZE = 4
TIMESTEPS = 1000

dataset = tf.data.Dataset.from_tensor_slices(sketch_paths)
dataset = dataset.map(load_and_preprocess, num_parallel_calls=tf.data.AUTOTUNE)
dataset = dataset.shuffle(100).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

# ==========================================
# 2. THE DIFFUSION MATH (FORWARD PROCESS)
# ==========================================
# Linear variance schedule
beta = tf.linspace(0.0001, 0.02, TIMESTEPS)
alpha = 1.0 - beta
alpha_bar = tf.math.cumprod(alpha)

def forward_diffusion(x_0, t):
    """Jumps to step 't' and adds the exact right amount of noise."""
    a_bar_t = tf.gather(alpha_bar, t)
    a_bar_t = tf.reshape(a_bar_t, (-1, 1, 1, 1))
    
    noise = tf.random.normal(shape=tf.shape(x_0))
    mean = tf.sqrt(a_bar_t) * x_0
    variance = tf.sqrt(1.0 - a_bar_t) * noise
    
    return mean + variance, noise

# ==========================================
# 3. THE TIME-CONDITIONED U-NET
# ==========================================
def build_diffusion_unet():
    image_input = layers.Input(shape=[64, 64, 3], name="noisy_image")
    time_input = layers.Input(shape=[1], name="time_step")

    # Time Embedding
    t_emb = layers.Dense(128, activation='swish')(time_input)
    t_emb = layers.Dense(128, activation='swish')(t_emb)

    # ENCODER
    e1 = layers.Conv2D(64, 4, strides=2, padding='same')(image_input)
    time_proj_1 = layers.Reshape((1, 1, 64))(layers.Dense(64)(t_emb))
    e1 = layers.Add()([e1, time_proj_1])
    e1 = layers.LeakyReLU()(e1)

    e2 = layers.Conv2D(128, 4, strides=2, padding='same')(e1)
    e2 = layers.BatchNormalization()(e2)
    time_proj_2 = layers.Reshape((1, 1, 128))(layers.Dense(128)(t_emb))
    e2 = layers.Add()([e2, time_proj_2])
    e2 = layers.LeakyReLU()(e2)

    e3 = layers.Conv2D(256, 4, strides=2, padding='same')(e2)
    e3 = layers.BatchNormalization()(e3)
    e3 = layers.LeakyReLU()(e3)

    # DECODER
    d1 = layers.Conv2DTranspose(128, 4, strides=2, padding='same')(e3)
    d1 = layers.BatchNormalization()(d1)
    d1 = layers.ReLU()(d1)
    d1 = layers.Concatenate()([d1, e2]) 
    
    d2 = layers.Conv2DTranspose(64, 4, strides=2, padding='same')(d1)
    d2 = layers.BatchNormalization()(d2)
    d2 = layers.ReLU()(d2)
    d2 = layers.Concatenate()([d2, e1]) 
    
    # Linear output predicting the noise, not pixels
    last = layers.Conv2DTranspose(3, 4, strides=2, padding='same')(d2)
    return Model(inputs=[image_input, time_input], outputs=last)

unet = build_diffusion_unet()

# ==========================================
# 4. TRAINING LOOP (PREDICTING NOISE)
# ==========================================
optimizer = tf.keras.optimizers.Adam(learning_rate=2e-4)
mse_loss = tf.keras.losses.MeanSquaredError()

@tf.function
def train_step(real_images):
    # 1. Pick a random time step for each image in the batch
    batch_size = tf.shape(real_images)[0]
    t = tf.random.uniform(shape=(batch_size,), minval=0, maxval=TIMESTEPS, dtype=tf.int32)
    
    # 2. Destroy the data (Forward Process)
    noisy_images, actual_noise = forward_diffusion(real_images, t)
    
    with tf.GradientTape() as tape:
        # 3. Ask the U-Net to predict the noise
        # We must cast 't' to float32 so it can pass through the Dense layers
        predicted_noise = unet([noisy_images, tf.cast(t, tf.float32)], training=True)
        
        # 4. Loss is simply how wrong the U-Net was about the noise
        loss = mse_loss(actual_noise, predicted_noise)
        
    gradients = tape.gradient(loss, unet.trainable_variables)
    optimizer.apply_gradients(zip(gradients, unet.trainable_variables))
    return loss

# ==========================================
# 5. REVERSE PROCESS (THE MAGIC)
# ==========================================
def generate_sketches(model, num_images=3):
    """Starts with pure noise and walks backward through time."""
    # 1. Start with pure random static
    x_t = tf.random.normal((num_images, 64, 64, 3))
    
    # 2. Walk backward from T=999 down to 0
    for t in reversed(range(0, TIMESTEPS)):
        t_batch = tf.fill([num_images], tf.cast(t, tf.float32))
        
        # Ask U-Net to guess the noise at this specific time step
        predicted_noise = model([x_t, t_batch], training=False)
        
        # Fetch the math constants for this step
        alpha_t = alpha[t]
        a_bar_t = alpha_bar[t]
        beta_t = beta[t]
        
        # The Reverse DDPM Formula: Subtract the predicted noise mathematically
        if t > 0:
            noise = tf.random.normal(shape=tf.shape(x_t))
        else:
            noise = tf.zeros_like(x_t) # No noise added on the very last step
            
        x_t = (1 / tf.sqrt(alpha_t)) * (x_t - ((1 - alpha_t) / tf.sqrt(1 - a_bar_t)) * predicted_noise) + tf.sqrt(beta_t) * noise
        
    # 3. Display the final carved images
    plt.figure(figsize=(9, 3))
    for i in range(num_images):
        plt.subplot(1, num_images, i+1)
        display_img = x_t[i] * 0.5 + 0.5 # Rescale to [0,1]
        plt.imshow(tf.clip_by_value(display_img, 0.0, 1.0))
        plt.axis('off')
    plt.show()

# ==========================================
# 6. EXECUTE
# ==========================================
def train(dataset, epochs=100):
    for epoch in range(epochs):
        start = time.time()
        total_loss = 0
        batch_count = 0
        
        for image_batch in dataset:
            loss = train_step(image_batch)
            total_loss += loss
            batch_count += 1
            
        display.clear_output(wait=True)`
        print(f'Epoch {epoch+1} | Average Loss: {total_loss/batch_count:.4f} | Time: {time.time()-start:.2f} sec')
        
        # Generate samples every 10 epochs to watch it learn
        if (epoch + 1) % 10 == 0:
            generate_sketches(unet)

print("Igniting Diffusion Engine...")
train(dataset, epochs=100)
    
```
